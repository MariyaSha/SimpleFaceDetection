# Face and Object Detection in Live Videos

## Detect Objects with Official YOLO Model

The following code will detect different objects in live video feeds such as: person, cat, laptop, cup, etc.

### Camera Integration
Please make sure you have an integrated webcam or some external camera connected to your computer before running this. I'm using DJI Pocket Osmo 3 for this workflow, and therefore `cv2.VideoCapture(2)` is set to the camera at index `2`. Integrated cameras on the other hand, are using index `0` such that `cv2.VideoCapture(0)`. Otherwise, you can try different numbers until you find the correct camera.

### Other Resources
No need to download additional resources - YOLO will be automatically downloaded from ultralytics with `YOLO("yolov8n.pt")`.

````python
import cv2
from ultralytics import YOLO

# capture video feed from camera
cap = cv2.VideoCapture(2)

# install and load YOLO model
model = YOLO("yolov8n.pt")

# display video feed until the "x" key is pressed on the keyboard
while cv2.waitKey(1) != ord("x"):
    _, frame = cap.read()

    # send each frame to the model
    result = model(frame, verbose=False)

    # display the object detection results
    processed_feed = result[0].plot()
    cv2.imshow("my window", processed_feed)

    # [optional] display window in a specific location on the monitor
    #cv2.moveWindow("my window", 1800, 200)

cv2.waitKey(5000)
cap.release()
cv2.destroyAllWindows()
````

## Detect Faces with Community YOLO-Face Model

For this you'll need to download `yolov8m-face.pt` from:
https://github.com/akanametov/yolo-face

Make sure you save it in the same folder as this notebook - and you're good to go.

````python
import cv2
from ultralytics import YOLO

# capture video feed from camera
cap = cv2.VideoCapture(2)

# load YOLO-Face model from local folder
face_model = YOLO("yolov8m-face.pt")

# display video feed until the "x" key is pressed on the keyboard
while cv2.waitKey(1) != ord("x"):
    _, frame = cap.read()

    # send each frame to the model
    face_result = face_model(frame, verbose=False)

    # display the face detection results
    processed_feed = face_result[0].plot()
    cv2.imshow("my window", processed_feed)

    # [optional] display window in a specific location on the monitor
    #cv2.moveWindow("my window", 1800, 200)

cv2.waitKey(5000)
cap.release()
cv2.destroyAllWindows()
````

## Detect Objects and Faces, Combining Two YOLO Models
The following code combines both models, working on the same image.
Make sure you doanloaded `yolov8m-face.pt` from the previous section before running it.

Please note: the following workflow is not optimal for CPU, you'll need to run it on GPU instead to make it work smoothly. 

````python
import cv2
from ultralytics import YOLO

# capture video feed from camera
cap = cv2.VideoCapture(2)

# load models
model = YOLO("yolov8n.pt")
face_model = YOLO("yolov8m-face.pt")

# display video feed until the "x" key is pressed on the keyboard
while cv2.waitKey(1) != ord("x"):
    _, frame = cap.read()

    # send each frame to both models
    result = model(frame, verbose=False)
    face_result = face_model(frame, verbose=False)

    # display the face detection results
    object_detection_feed = result[0].plot()
    combined_feed = face_result[0].plot(img=object_detection_feed)
    cv2.imshow("my window", combined_feed)

    # [optional] display window in a specific location on the monitor
    #cv2.moveWindow("my window", 1800, 200)

cv2.waitKey(5000)
cap.release()
cv2.destroyAllWindows()
````

## Detect Objects and Faces with GPU

To run the following code, you'll need to set up GPU processing.
Currecntly, as of July 2026, you do so with:
```
pip uninstall torch torchvision
```

Followed by:
```
pip install torch torchvision --index-url https://download.pytorch.org/whl/cu130
```

Then you can run the next cell as is

````python
import cv2
from ultralytics import YOLO

# capture video feed from camera
cap = cv2.VideoCapture(2)

# load models
model = YOLO("yolov8n.pt").to("cuda")
face_model = YOLO("yolov8m-face.pt").to("cuda")

# display video feed until the "x" key is pressed on the keyboard
while cv2.waitKey(1) != ord("x"):
    _, frame = cap.read()

    # send each frame to both models
    result = model(frame, verbose=False)
    face_result = face_model(frame, verbose=False)

    # display the face detection results
    object_detection_feed = result[0].plot()
    combined_feed = face_result[0].plot(img=object_detection_feed)
    cv2.imshow("my window", combined_feed)

    # [optional] display window in a specific location on the monitor
    #cv2.moveWindow("my window", 1800, 200)

cv2.waitKey(5000)
cap.release()
cv2.destroyAllWindows()
````
