# Face and Object Detection in Images

## Detect Objects with Official YOLO Model

The following code will detect different objects in images such as: person, cat, laptop, cup, etc.

No need to download additional resources - YOLO will be automatically downloaded from ultralytics with `YOLO("yolov8n.pt")`.

```python
import cv2
from ultralytics import YOLO

# install and load YOLO model
model = YOLO("yolov8n.pt")

# send a demo image to the model
photo = cv2.imread("demo.jpg")
detection_result = model(photo)

# display the detection results
processed_image = detection_result[0].plot()
cv2.imshow("my window", processed_image)

# [optional] display window in a specific location on the monitor
#cv2.moveWindow("my window", 1800, 200)

cv2.waitKey(5000) # display window for 5 seconds
cv2.destroyAllWindows()
```

## Detect Faces with Community YOLO-Face Model

For this you'll need to download `yolov8m-face.pt` from:
https://github.com/akanametov/yolo-face

Make sure you save it in the same folder as this notebook - and you're good to go.

```python
import cv2
from ultralytics import YOLO

# load YOLO-Face model from local folder
face_model = YOLO("yolov8m-face.pt")

# send a demo image to the model
photo = cv2.imread("demo.jpg")
face_detection_result = face_model(photo)

# display the face detection results
processed_image = face_detection_result[0].plot()
cv2.imshow("my window", processed_image)

# [optional] display window in a specific location on the monitor
#cv2.moveWindow("my window", 1800, 200)

cv2.waitKey(5000) # display window for 5 seconds
cv2.destroyAllWindows()
```

## Detect Objects and Faces, Combining Two YOLO Models
The following code combines both models, working on the same image.
Make sure you doanloaded `yolov8m-face.pt` from the previous section before running it.

```python
import cv2
from ultralytics import YOLO

# load both models
model = YOLO("yolov8n.pt")
face_model = YOLO("yolov8m-face.pt")

# send the same demo image to both models
photo = cv2.imread("demo.jpg")
result = model(photo)
face_result = face_model(photo)

# display combined detection results in the same window
cat_person = result[0].plot()
cat_person_face = face_result[0].plot(img=cat_person)
cv2.imshow("my window", cat_person)

# [optional] display window in a specific location on the monitor
#cv2.moveWindow("my window", 1800, 200)

cv2.waitKey(5000)
cv2.destroyAllWindows()
```
